```python
import random

input_file = open("/content/INPUT.txt", 'r')
first_line = input_file.readline()
numbers = first_line.split(' ')
num_courses = int(numbers[0])
num_timeslots = int(numbers[1])
chromosome_size = num_timeslots * num_courses
pop_size = 10

def initialize_population(pop_size, num_courses, num_timeslots):
    chromosome_length = num_courses * num_timeslots
    return [''.join(random.choice('01') for i in range(chromosome_length)) for i in range(pop_size)]

def fitness_function(chromosome, num_courses, num_timeslots):
    pen_over = 0
    pen_con = 0

    for i in range(num_timeslots):
        timeslot = chromosome[i * num_courses:(i + 1) * num_courses]
        scheduled_courses = timeslot.count('1')
        if scheduled_courses > 1:
            pen_over += (scheduled_courses - 1)

    for i in range(num_courses):
        Scount = sum(chromosome[j * num_courses + i] == '1' for j in range(num_timeslots))
        if Scount != 1:
            pen_con += abs(Scount - 1)

    total_penalty = pen_over + pen_con
    return -total_penalty

def select_parents(population):
    return random.sample(population, 2)


def crossover(parent_1, parent_2):
    cpoint = random.randint(0, len(parent_1) - 1)

    offspring1 = parent_1[:cpoint] + parent_2[cpoint:]
    offspring2 = parent_2[:cpoint] + parent_1[cpoint:]
    return offspring1, offspring2


def mutatation(chromosome, mut_rate):
    return ''.join(
        gene if random.random() > mut_rate else str(1 - int(gene))
        for gene in chromosome
    )


def gen_algorithm(num_courses, num_timeslots, max_gen=100, population_size=10, mut_rate=0.01):
    population = initialize_population(population_size, num_courses, num_timeslots)
    op_fitness = float('-inf')
    op_chromosome = None

    for generation in range(max_gen):
        population_new = []
        for i in range(population_size // 2):
            parent_1, parent_2 = select_parents(population)
            offspring1, offspring2 = crossover(parent_1, parent_2)
            offspring1 = mutatation(offspring1, mut_rate)
            offspring2 = mutatation(offspring2, mut_rate)
            population_new.extend([offspring1, offspring2])

        population = population_new

        for chromosome in population:
            fitness = fitness_function(chromosome, num_courses, num_timeslots)
            if fitness > op_fitness:
                op_fitness = fitness
                op_chromosome = chromosome

    return op_chromosome, op_fitness

op_chromosome, op_fitness = gen_algorithm(num_courses, num_timeslots)

print(f"{op_chromosome}")
print(f"{op_fitness}")
```

```python
import random

def population_ini(population_size, chromosome_len):
    population = []

    for i in range(population_size):
        achromosome = ''.join(random.choice('01') for i in range(chromosome_len))
        population.append(achromosome)
    return population

def two_point(parent1, parent2):

    crossover1 = random.randint(0, len(parent1) - 2)
    crossover2 = random.randint(crossover1 + 1, len(parent1) - 1)
    print(f"Crossover points: {crossover1}, {crossover2}")

    offspring1 = parent1[:crossover1] + parent2[crossover1:crossover2] + parent1[crossover2:]
    offspring2 = parent2[:crossover1] + parent1[crossover1:crossover2] + parent2[crossover2:]

    return offspring1, offspring2

pop_size = 10
chrom_length = 9

population = population_ini(pop_size, chrom_length)

parent1 = random.choice(population)
parent2 = random.choice(population)

print(f"\nParent 1: {parent1}")
print(f"Parent 2: {parent2}")

offspring1, offspring2 = two_point(parent1, parent2)

print(f"\nOffspring 1: {offspring1}")
print(f"Offspring 2: {offspring2}")
```
