```python
def pacman_game(c):
    leaf = [3, 6, 2, 3, 7, 1, 2, 0]

    def minmax(depth, idx, maximizer, maxi, mini):
        if depth == 3:
            return leaf[idx]

        if maximizer:
            max_cal = float('-inf')
            for j in range(2):
                cal = minmax(depth + 1, idx * 2 + j, False, maxi, mini)
                max_cal = max(max_cal, cal)
                maxi = max(maxi, cal)
                if mini <= maxi:
                    break
            return max_cal
        else:
            min_cal = float('inf')
            for j in range(2):
                cal = minmax(depth + 1, idx * 2 + j, True, maxi, mini)
                min_cal = min(min_cal, cal)
                mini = min(mini, cal)
                if mini <= maxi:
                    break
            return min_cal

    no_magic = minmax(0, 0, True, float('-inf'), float('inf'))

    left_tree = max(leaf[0:4])
    right_tree = max(leaf[4:8])

    left_magic = left_tree - c
    right_magic = right_tree - c

    if right_magic > left_magic:
        direction_magic = "right"
        magic_val = right_magic
    else:
        direction_magic = "left"
        magic_val = left_magic

    if magic_val > no_magic:
        return f"The new minimax value is {magic_val}. Pacman goes {direction_magic} and uses dark magic."
    else:
        return f"The minimax value is {no_magic}. Pacman does not use dark magic."

cost = int(input("Enter the cost of using dark magic : "))
print(pacman_game(cost))
```
