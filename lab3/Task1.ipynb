```python
import random
class Node:
    def __init__(self, depth, player_user, utility_val=None):
        self.depth = depth
        self.player_user = player_user
        self.utility_val = utility_val
        self.branches = []

    def new_terminal(self):
        return self.depth == 0 or self.utility_val is not None

    def child_sum(self, child):
        self.branches.append(child)

def game_tree(depth, player_user):
    if depth == 0:
        utility_val = random.choice([-1, 1])
        return Node(depth, player_user, utility_val)

    node = Node(depth, player_user)
    for _ in range(2):
        child = game_tree(depth - 1, 1 - player_user)
        node.child_sum(child)
    return node

def alpha_beta_pruning(node, maxi, mini, maximizing_player):
    if node.new_terminal():
        return node.utility_val

    if maximizing_player:
        max_eval = float('-inf')
        for child in node.branches:
            eval = alpha_beta_pruning(child, maxi, mini, False)
            max_eval = max(max_eval, eval)
            maxi = max(maxi, eval)
            if mini <= maxi:
                break
        return max_eval
    else:
        min_eval = float('inf')
        for child in node.branches:
            eval = alpha_beta_pruning(child, maxi, mini, True)
            min_eval = min(min_eval, eval)
            mini = min(mini, eval)
            if mini <= maxi:
                break
        return min_eval

def mortal_kombat(start_play):
    rounds = []
    scorpion_wins = 0
    sub_zero_wins = 0

    for round_num in range(1, 4):
        base = game_tree(5, start_play)
        result = alpha_beta_pruning(base, float('-inf'), float('inf'), True)
        win = "Sub-Zero" if result == 1 else "Scorpion"
        rounds.append(f"Winner of Round {round_num}: {win}")
        if win == "Sub-Zero":
            sub_zero_wins += 1
        else:
            scorpion_wins += 1
        start_play = 1 - start_play

    game_winner = "Sub-Zero" if sub_zero_wins > scorpion_wins else "Scorpion"
    print(f"Game Winner: {game_winner}")
    print(f"Total Rounds Played: {len(rounds)}")
    for round_result in rounds:
        print(round_result)

start_play = int(input("Enter starting player (0 for Scorpion, 1 for Sub-Zero): "))
mortal_kombat(start_play)
```
